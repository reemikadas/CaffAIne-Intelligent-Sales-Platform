```sql
-- Create the CaffAIne catalog
CREATE CATALOG IF NOT EXISTS caffaine
COMMENT 'Data and AI assets for the CaffAIne sales intelligence platform';

-- Create the Bronze Schema
CREATE SCHEMA IF NOT EXISTS caffaine.bronze
COMMENT 'Raw ingested sales data and source-file storage';

-- Create a managed volume for monthly files
CREATE VOLUME IF NOT EXISTS caffaine.bronze.source_files
COMMENT 'Landing location for monthly coffee sales CSV files';

-- Create the Silver Schema
CREATE SCHEMA IF NOT EXISTS caffaine.silver
COMMENT 'Cleaned, validated, and conformed coffee sales data';
```
