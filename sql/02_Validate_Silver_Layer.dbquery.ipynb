```sql
-- Verify the Silver row count
SELECT
    COUNT(*) AS silver_row_count
FROM caffaine.silver.coffee_sales_clean;

-- Inspect the Silver schema
DESCRIBE TABLE caffaine.silver.coffee_sales_clean;

-- Inspect the Data Quality Rules
SELECT
    COUNT_IF(transaction_id IS NULL) AS null_transaction_id,
    COUNT_IF(transaction_timestamp IS NULL) AS null_transaction_timestamp,
    COUNT_IF(sales_amount IS NULL) AS null_sales_amount,
    COUNT_IF(
        sales_amount != CAST(transaction_qty * unit_price AS DECIMAL(12,2))
    ) AS incorrect_sales_amount
FROM caffaine.silver.coffee_sales_clean;

-- Inspect for duplicate records
SELECT
    COUNT(*) AS total_rows,
    COUNT(DISTINCT transaction_id) AS unique_transaction_ids,
    COUNT(*) - COUNT(DISTINCT transaction_id) AS duplication_transaction_ids
FROM caffaine.silver.coffee_sales_clean;
```
